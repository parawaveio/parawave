```python
import parawave
print('parawave imported successfully')
```

```python
@parawave(progress=None)
async def double(x: int) -> int:
    return x * 2

result = double.run(data=[{"x": 1}, {"x": 2}, {"x": 3}])
print(result.summary)
assert result.ok
assert len(result) == 3
print('Basic run test passed!')
```

```python
import json
j = result.to_json()
parsed = json.loads(j)
assert parsed['num_completed'] == 3
print('Export test passed!')
```
